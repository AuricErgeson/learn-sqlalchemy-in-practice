```python
from models import Product
from db import  Session
from sqlalchemy import select, or_, func, distinct
```

```python
session = Session ()
```

```python
session.get(Product,7)
```

### The first three products in alphabetical order built in the year 1983.


```python
query_1 = (select(Product)
           .order_by(Product.name)
           .where(Product.year == 1983)
           .limit(3))
session.scalars(query_1).all()
```

### Products that use the "Z80" CPU or any of its clones. Assume that all products based on this CPU have the word "Z80" in the cpu column.

```python
query_2 = select(Product).where(Product.cpu.like('%Z80%'))
session.scalars(query_2).all()
```

### Products that use either the "Z80" or the "6502" CPUs, or any of its clones, built before 1990, sorted alphabetically by name.

```python
query_3 = (select(Product)
           .where(or_(Product.cpu.like('%Z80%'),Product.cpu.like('%6502%')))
           .where(Product.year < 1990)
           .order_by(Product.name)
           )

session.scalars(query_3).all()
```

### The manufacturers that built products in the 1980s.

```python
query_4 = select(Product.manufacturer).where(Product.year == 1980)
session.scalars(query_4).all()
```

### Manufacturers whose names start with the letter "T", sorted alphabetically

```python
query_5 = select(Product.name).where(Product.name.like('T%')).order_by(Product.name)
session.scalars(query_5).all()
```

### The first and last years in which products have been built in Croatia, along with the number of products built.


```python
query_6 = (
    select(Product.country,
           func.min(Product.year),
           func.max(Product.year),
           func.count())
    .group_by(Product.country)
    .having(Product.country == 'Croatia')
)

session.execute(query_6).all()
```

#### The number of products that were built each year. The results should start
#### from the year with the most products, to the year with the least. Years in
#### which no products were built do not need to be included.

```python
query_7 =(
    select(Product.year,
           func.count())
    .group_by(Product.year)
    .order_by(func.count().desc())
)

session.execute(query_7).all()
```

### The number of manufacturers in the United States (note that the country field for these products is set to USA)

```python
query_8 = (
    select(func.count(Product.manufacturer.distinct()))
    .where(Product.country == 'USA')
)

session.scalar(query_8)
```
